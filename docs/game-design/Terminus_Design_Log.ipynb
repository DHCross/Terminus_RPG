# Terminus RPG Design Log
**Project**: Terminus RPG
**Rulebase**: `Terminus RPG 5.9.26.md`
**Purpose**: To track design decisions, mechanic iterations, open questions, and overall design trajectory as a living dashboard.

```
import pandas as pd
from datetime import date
import os

# Initialize or load the Design Log database
log_file = 'design_log.csv'
if not os.path.exists(log_file):
    df = pd.DataFrame(columns=['Date', 'Mechanic', 'Version', 'Change', 'Status', 'Rationale', 'Next_Steps'])
    df.to_csv(log_file, index=False)
    print(f'Created new design log: {log_file}')
else:
    df = pd.read_csv(log_file)
    print(f'Loaded design log with {len(df)} entries.')
```

## 1. Table of Contents Generator
Run this cell to generate a clickable Table of Contents from the current rules markdown file.

```
import re

rule_file = 'Terminus RPG 5.9.26.md'
with open(rule_file, 'r', encoding='utf-8') as f:
    lines = f.readlines()
    
toc = []
for i, line in enumerate(lines, 1):
    if line.startswith('## '):
        title = line.strip('# \n')
        toc.append(f'- [{title}](#L{i})')
print('\n'.join(toc))
```

## 2. Log a New Design Decision
Update the dictionary below to track a new decision.

```
new_entry = {
    'Date': date.today().isoformat(),
    'Mechanic': 'Orders & Asymmetry',
    'Version': '5.9.26',
    'Change': 'Added Approaches & Signatures, clarified Exert tension.',
    'Status': 'Implemented',
    'Rationale': 'Orders felt too balanced/safe. Need more texture and specific narrative permissions.',
    'Next_Steps': 'Draft Section 15: Magic & Civic Workings to tackle the magic system gap.'
}

df = pd.read_csv(log_file)
df = pd.concat([df, pd.DataFrame([new_entry])], ignore_index=True)
df.to_csv(log_file, index=False)
print('Logged new decision.')
df.tail(3)
```
